```python
df['timeStamp'].iloc[0],type(df['timeStamp'].iloc[0]) # Here time is in string datatype so we have to convert it.
```

```python
df['timeStamp']= pd.to_datetime(df['timeStamp'])
```

```python
time=df['timeStamp'].iloc[0]
type(time),time, time.month, time.year
```

```python
# Creating new coulumns
df['year'] = df['timeStamp'].apply(lambda time: time.year)
df['month'] = df['timeStamp'].apply(lambda time: time.month)
df['date'] = df['timeStamp'].apply(lambda time: time.date())
df['hour'] = df['timeStamp'].apply(lambda time: time.hour)
df['week_day'] = df['timeStamp'].apply(lambda time: time.dayofweek)
df.head(2)
```

```python
# In the place of day and month, we are representing their respective names using map()
day_map={0:'sun',1:'Mon',2:'Tue',3:'Wed',4:'Thu',5:'Fri',6:'Sat'}
month_map={1:'Jan',2:'Feb',3:'Mar',4:'Apr',5:'May',6:'Jun',7:'Jul',8:'Aug',9:'Sep',10:'Oct',11:'Nov',12:'Dec'}

df['day_name']=df['week_day'].map(day_map)
df['month_name']=df['month'].map(month_map)
df.head(2)
```

```python
df['month_name'].value_counts().plot(kind='bar') 
```

```python
sns.countplot(x='day_name',data=df,hue='Reason',palette='deep')
#To relocate the legend 
plt.legend(bbox_to_anchor=(1,1),loc=2)
```

```python
sns.countplot(x='month_name',data=df,hue='Reason',palette='Set1')
#To relocate the legend 
plt.legend(bbox_to_anchor=(1,1),loc=2)
```

```python
# we can fill these missing info byb using line plot
# group data by month and aggregate it by count()
byMonth= df.groupby('month').count()
byMonth.head(12)
```

```python
byMonth['zip'].plot()
```

```python
data = byMonth.reset_index()
data.head()
```

```python
sns.lmplot(x='month',y='twp',data=data)
```

```python
#calls per day on specific reasons(fire,traffic,ems) using date 
n_calls_byData_Traffic=df[df['Reason']=='Traffic'].groupby('date').count()
n_calls_byData_Fire=df[df['Reason']=='Fire'].groupby('date').count()
n_calls_byData_EMS=df[df['Reason']=='EMS'].groupby('date').count()
```

```python
fig,(ax1,ax2,ax3)=plt.subplots(nrows=3,figsize=(8,6),sharex=True,sharey=True)

n_calls_byData_Traffic['zip'].plot(ax=ax1)
ax1.set_title('Traffic')

n_calls_byData_Fire['zip'].plot(ax=ax2)
ax1.set_title('Fire')

n_calls_byData_EMS['zip'].plot(ax=ax3)
ax1.set_title('EMS')

plt.tight_layout()
```
